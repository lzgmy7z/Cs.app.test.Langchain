```python
from dotenv import load_dotenv
load_dotenv()

from langchain.chat_models import init_chat_model
import os

model = init_chat_model(
    model = 'qwen3.5-puls',
    model_provider = 'openai',
    base_url =os.getenv('DASHSCOPE_BASE_URL'),
    api_url = os.getenv('DASHSCOPE_API_KEY'),

)
```

```python
from langchain_tavily import TavilySearch
web_sh = TavilySearch(
    max_results = 5,
    topic = 'general',
)
```

```python
from langgraph.checkpoint.sqlite import SqliteSaver
import sqlite3

connection = sqlite3.connect('rs/ps_c.db', check_same_thread = False)
checkp = SqliteSaver(connection)
checkp.setup()
```

```python
from langchain.agents import create_agent
sys_p = "你是坤坤"

agent = create_agent(
    model = model,
    tools = [web_sh],
    system_prompt = sys_p,
    checkpointer = checkp,
)
```

```python
from langchain.messages import HumanMessage

mm_m = HumanMessage([
    {'type':'text',"text":"给我挑一个"},
    {"type":'image',"url":"--"},
])
config = {'configurable':{"thread_id":"thread_1"}}

resp = agent.invoke({"messages":[mm_m]},config)
for message in resp["messages"]:
    message.pretty_print()
```

```python
resp = agent.invoke({
    'messages':[HumanMessage(contont="说详细点")]},config
    )
for message in resp['messages']:
    message.pretty_print()
```
